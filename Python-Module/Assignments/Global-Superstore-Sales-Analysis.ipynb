## Final Business Summary

The Global Superstore analysis shows that the business generated total sales of approximately **12.64 million** and total profit of approximately **1.47 million**, with **178,312 units** sold. The overall average discount was **14.29%**, while total shipping costs amounted to approximately **1.36 million**.

### Key Business Insights

**1. Market Performance**  
Europe was the most profitable market, generating approximately **449,552 in profit**, despite having the lowest average discount of **9.09%**. Asia Pacific generated the highest sales at approximately **4.04 million**, but its profit was lower than Europe's at approximately **403,176**. Asia Pacific also had the highest average discount at **18.10%**. This suggests that higher sales do not necessarily result in higher profits and that excessive discounting may reduce profitability.

**2. Category Performance**  
Technology was the most profitable category, generating approximately **663,779 in profit** from **4.74 million in sales**. Office Supplies had the highest sales volume, with **108,244 units sold**, but generated a lower profit of approximately **518,596**. Furniture generated approximately **4.11 million in sales** but had the lowest profit at approximately **285,083**. This indicates that sales volume alone does not determine profitability.

**3. Overall Business Performance**  
The analysis indicates that the business performs strongly in terms of sales and profit, but there are differences in profitability across markets and product categories. Markets and categories with higher sales should not automatically be considered the most successful because discounts, costs, and profit margins also affect overall performance.

### Recommendations

- **Review discounting strategies**, particularly in markets such as Asia Pacific where the average discount is relatively high.
- **Prioritize Technology products**, as they generate the highest total profit.
- **Investigate Furniture profitability** to identify whether high costs, discounts, or low margins are reducing profits.
- **Focus on profitability rather than sales volume alone** when evaluating markets and product categories.
- **Monitor shipping costs** and shipping duration to identify opportunities to reduce operational costs and improve efficiency.
