# Project 1: NumPy — Real Estate Market Analysis

**Role:** Data Analyst, Real Estate Company
**Goal:** Analyze property data using NumPy to identify patterns in property
prices and characteristics.

**Dataset:** [Housing Prices Regression Dataset](https://www.kaggle.com/datasets/denkuznetz/housing-prices-regression) (Kaggle)

**Concepts covered:** NumPy arrays, structured arrays, indexing, aggregation,
Boolean masking, broadcasting, sorting, and fancy indexing.


## 1. Load and Explore the Data
**Concepts: Structured Arrays, Indexing**

```python
import numpy as np

# Load the CSV as a NumPy structured array.
# names=True reads the header row as field names.
# dtype=None lets NumPy infer each column's data type.
# encoding=None keeps text fields as regular Python strings.
data = np.genfromtxt("real_estate_dataset.csv", delimiter=",",
                      names=True, dtype=None, encoding=None)

print("Shape:", data.shape)
print("\nField names:", data.dtype.names)
print("\nField data types:")
for name in data.dtype.names:
    print(f"  {name}: {data.dtype[name]}")
```

```python
# Display the first five records
print(data[:5])
```

**Interpretation:** The dataset loads as a 1-dimensional structured array,
where each element is one property (one row), and each field (column) can be
accessed by name, e.g. `data['Price']`. This is the NumPy equivalent of a
table, without needing pandas.

## 2. Analyze Property Prices
**Concepts: Indexing, Aggregation Functions**

```python
prices = data['Price']

price_mean = np.mean(prices)
price_median = np.median(prices)
price_min = np.min(prices)
price_max = np.max(prices)
price_p25 = np.percentile(prices, 25)
price_p75 = np.percentile(prices, 75)

print(f"Mean price:   ${price_mean:,.2f}")
print(f"Median price: ${price_median:,.2f}")
print(f"Min price:    ${price_min:,.2f}")
print(f"Max price:    ${price_max:,.2f}")
print(f"25th percentile: ${price_p25:,.2f}")
print(f"75th percentile: ${price_p75:,.2f}")
```

**Interpretation:** The mean price ($582,210) is close to the
median ($574,724), suggesting prices are fairly evenly
distributed rather than skewed by a handful of extreme outliers. The
25th–75th percentile range shows where the "middle half" of the market
sits.

## 3. Analyze Property Size
**Concepts: Indexing, Aggregation Functions**

```python
sizes = data['Square_Feet']

size_mean = np.mean(sizes)
size_median = np.median(sizes)
size_min = np.min(sizes)
size_max = np.max(sizes)

print(f"Mean size:   {size_mean:,.1f} sq ft")
print(f"Median size: {size_median:,.1f} sq ft")
print(f"Min size:    {size_min:,.1f} sq ft")
print(f"Max size:    {size_max:,.1f} sq ft")
```

**Interpretation:** The gap between minimum and maximum size shows how
diverse the properties in this dataset are, from small starter homes to large
estates. A mean close to the median suggests sizes are fairly evenly
distributed, without extreme outliers dominating the average.

## 4. Calculate Price per Square Foot
**Concept: Element-wise Array Operations**

```python
# Element-wise division: each property's price divided by its own size
price_per_sqft = data['Price'] / data['Square_Feet']

print(f"Mean price/sqft:   ${np.mean(price_per_sqft):,.2f}")
print(f"Median price/sqft: ${np.median(price_per_sqft):,.2f}")
print(f"Min price/sqft:    ${np.min(price_per_sqft):,.2f}")
print(f"Max price/sqft:    ${np.max(price_per_sqft):,.2f}")
```

**Interpretation:** Price per square foot standardizes properties of
different sizes so they can be compared fairly. A wide range between the
minimum and maximum suggests that factors other than size alone — like
location — are driving a large part of the price.

## 5. Identify Expensive Properties
**Concepts: Boolean Masking, Aggregation Functions**

```python
# Boolean mask: True for properties priced in the top 25% (at or above the 75th percentile)
top25_mask = data['Price'] >= price_p75
expensive_properties = data[top25_mask]

n_expensive = len(expensive_properties)
avg_expensive_price = np.mean(expensive_properties['Price'])
overall_avg_price = np.mean(data['Price'])
difference = avg_expensive_price - overall_avg_price

print(f"Number of properties in top 25% by price: {n_expensive}")
print(f"Average price of top 25%:   ${avg_expensive_price:,.2f}")
print(f"Overall average price:      ${overall_avg_price:,.2f}")
print(f"Difference (top 25% vs overall): ${difference:,.2f}")
```

**Interpretation:** The top 25% of properties average
$744,143, about $161,933 above the overall
market average of $582,210 — confirming a clear premium segment
exists well above typical market pricing.

## 6. Apply Multiple Conditions
**Concepts: Boolean Masking, Multiple Boolean Conditions**

```python
avg_size = np.mean(data['Square_Feet'])

# Combine three Boolean conditions with & (all must be true)
combined_mask = (
    (data['Square_Feet'] > avg_size) &
    (data['Price'] >= price_p75) &
    (data['Location_Score'] >= 8)
)

premium_properties = data[combined_mask]

print(f"Properties meeting all 3 conditions: {len(premium_properties)}")
print("\nSample of matching properties:")
print(premium_properties[['Square_Feet', 'Price', 'Location_Score']][:5])
```

**Interpretation:** Only 17 properties (out of 500) satisfy
all three conditions at once — above-average size, top-quartile price, and a
location score of 8+. This is a small fraction of the market, indicating
that large size, high price, and prime location don't automatically come
together; properties combining all three represent a distinct, elite
segment.

## 7. Compare Prices to the Average
**Concept: Broadcasting**

```python
# Broadcasting: subtracting a single scalar (overall_avg_price) from every
# element in the prices array, without writing a loop
price_diff_from_avg = data['Price'] - overall_avg_price

above_avg_mask = price_diff_from_avg > 0
below_avg_mask = price_diff_from_avg < 0

print(f"Properties above average: {np.sum(above_avg_mask)}")
print(f"Properties below average: {np.sum(below_avg_mask)}")
print(f"\nExample differences (first 5 properties): {price_diff_from_avg[:5]}")
```

**How broadcasting was applied:** `overall_avg_price` is a single number
(a scalar), while `data['Price']` is an array of hundreds of values. NumPy
automatically "broadcasts" the single scalar across every element of the
array, subtracting it from each property's price in one operation, with no
explicit loop required.

## 8. Find the 10 Most Expensive Properties
**Concepts: Sorting, Fancy Indexing**

```python
# argsort returns the indices that would sort the array; [::-1] reverses
# it to descending order, and [:10] takes the top 10 indices
top10_idx = np.argsort(data['Price'])[::-1][:10]

# Fancy indexing: passing an array of indices to pull out multiple
# specific rows/fields at once
top10 = data[top10_idx]

print("10 Most Expensive Properties:")
print(top10[['ID', 'Square_Feet', 'Num_Bedrooms', 'Location_Score', 'Price']])
```

**Interpretation:** Sorting by price and using fancy indexing lets us
pull out the full records of the 10 highest-priced properties at once,
including every field, rather than just the price values alone.

## 9. Final Analysis

```python
top10_avg_size = np.mean(top10['Square_Feet'])
top10_avg_beds = np.mean(top10['Num_Bedrooms'])
top10_avg_location = np.mean(top10['Location_Score'])

overall_avg_size = np.mean(data['Square_Feet'])
overall_avg_beds = np.mean(data['Num_Bedrooms'])
overall_avg_location = np.mean(data['Location_Score'])

print(f"{'Metric':<20}{'Top 10':>12}{'Overall Avg':>15}")
print(f"{'Size (sqft)':<20}{top10_avg_size:>12,.0f}{overall_avg_size:>15,.0f}")
print(f"{'Bedrooms':<20}{top10_avg_beds:>12.1f}{overall_avg_beds:>15.1f}")
print(f"{'Location Score':<20}{top10_avg_location:>12.1f}{overall_avg_location:>15.1f}")
```

**What characteristics tend to be associated with the most expensive
properties in this dataset?**

The 10 most expensive properties average 267 sq ft versus a
dataset-wide average of 175 sq ft, 4.7 bedrooms versus
3.0 overall, and a location score of 6.6 versus
5.2 overall. All three metrics are higher among the top 10,
suggesting that size, bedroom count, and location quality are jointly
associated with premium pricing, with location score showing one of the
largest relative gaps.

**Business Insight:** Property size and location score are the two factors
most consistently elevated among the highest-priced homes in this dataset.
For a real estate company, this means marketing and pricing strategy for
premium listings should emphasize square footage and neighborhood quality
above secondary features like pool or garden presence. Investors looking to
maximize resale value should prioritize acquiring larger properties in
highly-rated locations, since these two traits — more than any single
amenity — appear to command the steepest price premiums in this
market.
