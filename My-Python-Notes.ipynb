### Calculator Assignment


```python
# The user is instructed to type in the first number, the number is converted to a float and stored in the num1 variable
num1 = float(input("Enter first number:"))
# The user is instructed to type in the second number, the number is converted to a float and stored in the num2 variable
num2 = float(input("Enter second number:"))
# User types in the desired operator
operator = input("Enter operator (+,-,*,/):")
if operator == "+":
   result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator =="*":
     result = num1 * num2
elif operator == "/":
 # The if statement here is not part of the major workflow    
     if num2 == 0:
         result = "Error:Division by Zero"
     else:
         result = num1/num2
else:
    result = "Invalid Operator!"
print(result)      
```
