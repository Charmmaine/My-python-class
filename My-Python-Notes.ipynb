### Sets
- Collection of values where each value is unique
- no duplicates, unordered, unindexed 
- created using curly braces or set()
- you can add or remove values but you cannot change them
- ideal to identify and remove duplicates
- its quick to search compared to other data structures such as lists

**creating a set**
- set use {}
- different from dictionaries as they do not include ":"
- no key value pair

```python
mahamri_ingredients = {"coconut milk", "cardamom", "sugar", "wheat flour","yeast","fresh palm wine","cooking oil"}
print(mahamri_ingredients)
type(mahamri_ingredients)
```

**Converting to a set**


```python
duplicate_ingredients = {"coconut milk", "cardamom", "sugar", "wheat flour","yeast","fresh palm wine","cooking oil","sugar"}
print(type(duplicate_ingredients))
print(duplicate_ingredients)
```

```python
unique_ingredients= set(duplicate_ingredients)
print(unique_ingredients)
type(unique_ingredients)
```

**limitations of sets**
- do not have an index meanung they are not ordered
- can't subset sets / slice them
- we cannot then have duplicates

```python
student_grades = [10,13,65,59,21,45,90,33,0,88]
```

```python
student_grades[2]
```

**Conditional Statements in For Loops**


```python
# Print out only grades that are greater than or equal to 50
for grade in student_grades:
    if grade >= 50:
        print(grade)
```

```python
print(student_grades)
```

```python
# Check how many students have scored 50 and above
graduating_students = 0
for grade in student_grades:
    if grade >= 50:
        graduating_students += 1
print(f"Number of students who have scored 50 and above: {graduating_students}")
```

```python
graduating_students = 0
students_not_graduating = 0
for grade in student_grades:
    if grade >= 50:
        graduating_students += 1
    else:
         students_not_graduating +=1
print(f"Number of students who are graduating: {graduating_students}")
print((f"Number of students who are not graduating: {students_not_graduating}"))
print(f"the total Number of students {graduating_students + students_not_graduating}")
```

**While loops**


- break and continue statements
- while loops are very useful continuous tasks:
   - Accelerate "while" a button is bein pressed 
   - Monitor while above or below threshhold
   

```python
#items in stock
stock = 10
# Number of purchases
purchases = 0
while purchases < stock:
    #when a customer buys something increase the number of purchases by 1
    purchases += 1
    #print for me the remaining stock after each purchase
    print(f"remaining stock: {stock - purchases}")
```
